```python
def GetMutProps(mutation):
    origres = mutation[0]
    newres = mutation[-1]
    position = mutation[1:-1]
    origprop = ResProps[origres]
    newprop = ResProps[newres]
    origVol, origCharge, origHydro, origHBond = origprop['Volume'], origprop['Charge'], origprop['Hydrophobicity'], origprop['HBond']
    newVol, newCharge, newHydro, newHBond = newprop['Volume'], newprop['Charge'], newprop['Hydrophobicity'], newprop['HBond']
    VolDif = newVol - origVol
    ChargeDif = newCharge - origCharge
    HydroDif = newHydro - origHydro

    if VolDif >= 20:
        VolChange = 'Larger Residue'
    elif VolDif <= -20:
        VolChange = 'Smaller Residue'
    else:
        VolChange = 'No Change'

    if ChargeDif > 0:
        ChargeChange = 'Positive Delta'
    elif ChargeDif < 0:
        ChargeChange = 'Negative Delta'
    else:
        ChargeChange = 'No Change'

    if HydroDif >= 1:
        HydroChange = 'More Hydrophobic'
    elif HydroDif <= -1:
        HydroChange = 'Less Hydrophobic'
    else:
        HydroChange = 'No Change'

    if (origHBond == 'yes' and newHBond == 'yes') or (origHBond == 'no' and newHBond == 'no'):
        HBondChange = 'No Change'
    elif origHBond == 'yes' and newHBond == 'no':
        HBondChange = 'Lose HBonds'
    elif origHBond == 'no' and newHBond == 'yes':
        HBondChange = 'Gain HBonds'
    return VolChange, ChargeChange, HydroChange, HBondChange

import numpy
from Bio.PDB.PDBParser import PDBParser
parser = PDBParser()
#structure = parser.get_structure("4OYY", "4oyy.pdb")
structure = parser.get_structure("4fnk", "4fnkA.pdb")

freqlist = []
vollist = []
chargelist = []
hydrolist = []
hbondlist = []
poslist = []
mutationlist = []
distlist = []
df1 = pd.DataFrame()
for mutation in newunimutlist:
    print(mutation)
    VolChange, ChargeChange, HydroChange, HBondChange = GetMutProps(mutation)
    vollist.append(VolChange)
    chargelist.append(ChargeChange)
    hydrolist.append(HydroChange)
    hbondlist.append(HBondChange)
    poslist.append(mutation[1:-1])
    mutationlist.append(mutation)
    if int(mutation[1:-1]) > 4 and int(mutation[1:-1]) < 195 :
        distlist.append(GetCatSerDist(structure, int(mutation[1:-1])))
    else:
        distlist.append(40.0)
df1['SizeDelta'] = vollist
df1['HydroDelta'] = hydrolist
df1['ChargeDelta'] = chargelist
df1['HBondDelta'] = hbondlist
df1['FreqPerc'] = df['freq'].astype('float')
df1['Position'] = poslist
df1['Mutation'] = mutationlist
df1['TriadDist'] = distlist

# if df['Sources']:
#     df1['Sources'] = df['Sources']
```
